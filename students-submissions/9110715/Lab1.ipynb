5. Create two 2x2 arrays and demonstrate how you can stack the elements vertically, horizontally, and split the arrays into smaller arrays.

```python
#Stacking and Splitting Arrays
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Vertical stacking (stacks along rows)
v_stacked = np.vstack((A, B))
print("Vertically Stacked:\n", v_stacked)

# Horizontal stacking (stacks along columns)
h_stacked = np.hstack((A, B))
print("Horizontally Stacked:\n", h_stacked)

# Split horizontally into 2 smaller arrays
split_arrays = np.hsplit(h_stacked, 2)
print("Split Array 1:\n", split_arrays[0])
print("Split Array 2:\n", split_arrays[1])
```

```python
X = np.array([[5, 7, 2], 
              [4, 5, 6], 
              [7, 4, 2]])  # Shape: (3, 3)

Y = np.array([[4, 2], 
              [6, 2], 
              [4, 2]])     # Shape: (3, 2)

# Is X @ Y possible? Yes! X has 3 columns, Y has 3 rows.
result = np.matmul(X, Y)  # or X @ Y
print("X * Y result:\n", result)



print("="*50)
# Demonstrating when it is NOT possible (Y @ X):
# Y has 2 columns, but X has 3 rows. This will raise a ValueError.
try:
    impossible_result = np.matmul(Y, X)
except ValueError as e:
    print("\nCannot multiply Y * X:", e)
```

8. How does broadcasting work? Demonstrate the subtraction, multiplication by considering a 3 x 3 matrix.

Broadcasting allows NumPy to perform element-wise arithmetic operations on arrays of different shapes (e.g., subtracting a vector/scalar from a matrix).

```python
M = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])  # 3x3 matrix

v = np.array([1, 2, 3])      # 1D array of shape (3,)

# Broadcasting Subtraction (subtracts v from each row of M)
sub_result = M - v
print("Broadcasting Subtraction:\n", sub_result)

# Broadcasting Multiplication (multiplies each row of M by v)
mult_result = M * v
print("\nBroadcasting Multiplication:\n", mult_result)
```

```python
import numpy as np

# --------------------------------------------------
# System 1:
# 2x1 + 3x2 - 4x3 = 6
# x1 - 4x2 + 0x3 = 8
# --------------------------------------------------

A1 = np.array([
    [2, 3, -4],
    [1, -4, 0]
], dtype=float)

b1 = np.array([6, 8], dtype=float)

# Find the rank of coefficient matrix and augmented matrix
rank_A1 = np.linalg.matrix_rank(A1)
rank_aug1 = np.linalg.matrix_rank(np.column_stack((A1, b1)))

print("System 1")
print("Coefficient matrix:")
print(A1)

print("\nRank of A:", rank_A1)
print("Rank of augmented matrix:", rank_aug1)

if rank_A1 == rank_aug1:
    if rank_A1 == A1.shape[1]:
        print("Number of solutions: One unique solution")
    else:
        print("Number of solutions: Infinitely many solutions")
else:
    print("Number of solutions: No solution")

# Solve using least squares to obtain a parametric solution
solution1 = np.linalg.lstsq(A1, b1, rcond=None)[0]

print("\nOne possible solution [x1, x2, x3]:")
print(solution1)


# --------------------------------------------------
# System 2:
# 3y1 - 4y2 + 5y3 = 10
# -y1 + 2y2 - 4y3 = 8
# --------------------------------------------------

A2 = np.array([
    [3, -4, 5],
    [-1, 2, -4]
], dtype=float)

b2 = np.array([10, 8], dtype=float)

# Find the rank of coefficient matrix and augmented matrix
rank_A2 = np.linalg.matrix_rank(A2)
rank_aug2 = np.linalg.matrix_rank(np.column_stack((A2, b2)))

print("\n\nSystem 2")
print("Coefficient matrix:")
print(A2)

print("\nRank of A:", rank_A2)
print("Rank of augmented matrix:", rank_aug2)

if rank_A2 == rank_aug2:
    if rank_A2 == A2.shape[1]:
        print("Number of solutions: One unique solution")
    else:
        print("Number of solutions: Infinitely many solutions")
else:
    print("Number of solutions: No solution")

# Solve using least squares to obtain one possible solution
solution2 = np.linalg.lstsq(A2, b2, rcond=None)[0]

print("\nOne possible solution [y1, y2, y3]:")
print(solution2)
```

```python
# System 2:
#  3*y1 - 4*y2 + 5*y3 = 10
# -1*y1 + 2*y2 - 4*y3 = 8

# Matrix A (Coefficients) and Vector b (Constants)
A2 = np.array([
    [ 3, -4,  5],
    [-1,  2, -4]
], dtype=float)

b2 = np.array([10, 8], dtype=float)

# 1. Determine number of solutions using matrix rank
rank_A2 = np.linalg.matrix_rank(A2)
augmented_A2 = np.column_stack((A2, b2))
rank_aug2 = np.linalg.matrix_rank(augmented_A2)
num_vars2 = A2.shape[1]

print("=== System 2 Analysis ===")
print(f"Rank of A: {rank_A2}")
print(f"Rank of Augmented Matrix [A|b]: {rank_aug2}")
print(f"Number of Variables: {num_vars2}")

if rank_A2 == rank_aug2:
    if rank_A2 == num_vars2:
        print("Number of solutions: Unique Solution")
    else:
        free_vars = num_vars2 - rank_A2
        print(f"Number of solutions: Infinitely Many Solutions ({free_vars} free variable)")
else:
    print("Number of solutions: No Solution")

# 2. Solve for y1, y2, y3 using Pseudoinverse (np.linalg.pinv)
y_sol = np.linalg.pinv(A2) @ b2

print("\nComputed Particular Solution:")
print(f"y1 = {y_sol[0]:.2f}")
print(f"y2 = {y_sol[1]:.2f}")
print(f"y3 = {y_sol[2]:.2f}")
```
